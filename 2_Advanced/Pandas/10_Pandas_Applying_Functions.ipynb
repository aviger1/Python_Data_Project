```python
#Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#Loading Dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_salary = df[pd.notna(df['salary_year_avg'])]
```

```python
def projected_salary(salary):
    return salary * 1.03
df_salary['df_salary_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)
df_salary[['salary_year_avg', 'df_salary_inflated']]

#df_salary['df_salary_inflated'] = df_salary['salary_year_avg'].apply(lambda x: x * 1.03)
```

```python
#Check if 'Senior' is present in the job title
def projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return row['salary_year_avg'] * 1.05
    else:
        return row['salary_year_avg'] * 1.03

df_salary['salary_year_inflated'] = df_salary.apply(projected_salary, axis=1)

df_salary[['job_title_short', 'salary_year_avg', 'salary_year_inflated']].groupby('job_title_short').mean().sort_values(by='salary_year_inflated', ascending=False).head(10)
```
